```python
import pandas as pd
import joblib
import sys
import os
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline


# 1. Cargar dataset

df = pd.read_csv("dataset.csv")

X = df["code"]
y = df["label"]

print(f"Dataset cargado: {len(df)} muestras")
print(f"Clases: {y.value_counts().to_dict()}")


# 2. Crear pipeline con TF-IDF

pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(max_features=5000)),
    ("rf", RandomForestClassifier(n_estimators=200, random_state=21))
])


# 3. Split de entrenamiento

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=21
)

print(f"\nTrain: {len(X_train)} | Test: {len(X_test)}")


# 4. Entrenar modelo

print("\nEntrenando modelo...")
pipeline.fit(X_train, y_train)


# 5. Evaluar

train_acc = pipeline.score(X_train, y_train)
test_acc = pipeline.score(X_test, y_test)

print(f"\nAccuracy en Train: {train_acc:.4f}")
print(f"Accuracy en Test: {test_acc:.4f}")


# 6. Guardar modelo

joblib.dump(pipeline, "model.pkl")

print("\n✓ Modelo guardado como model.pkl")
```
